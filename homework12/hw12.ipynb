## 2 Plotting
Make a scatter plot of **exoplanet Earth radius vs. equilibrium temperatures**. Include units, x/y axis labels, and a legend.

```python
# Your code here
import matplotlib.pyplot as plt

df = pd.read_csv('exoplanet_data.csv')

# x and y axis
x = df['pl_eqt']  # Equilibrium temperature (Kelvin)
y = df['pl_rade']  # Planet radius (Earth radius)

# Plotting
plt.figure(figsize=(8, 6))
plt.scatter(x, y, alpha=0.5, color='b', label="Exoplanets")
plt.xlabel("Equilibrium Temperature (K)")
plt.ylabel("Planet Radius (Earth Radius)")
plt.title("Exoplanet Radius vs. Equilibrium Temperature")
plt.legend()
plt.show()
```

What trend do you notice in the data? Does it have positive or negative correlation?

## 3 A "Cool" Planet
What planet has the **lowest** equilibrium temperature? Print the planet name and temperature with your `planet_data` variable.

```python
# Your code here
coldest_planet = planet_data.loc[planet_data['pl_eqt'].idxmin(), ['pl_name', 'pl_eqt']]
print(f'Planet: {coldest_planet['pl_name']}, Temperature: {coldest_planet['pl_eqt']}')
```

Make a new variable that only contains the information for that planet. \
\
*Hint: Use the `.iloc[[#]]` function. Replace # with the proper index.*

```python
# Your code here
coldest_index = planet_data['pl_eqt'].idxmin()
coldest_planet_data = planet_data.iloc[[coldest_index]]
print(coldest_planet_data)
```
